```python
def download_arrow(logical_key):
    file_hash = manifest[logical_key]
    url = f"{arrow_base_url}{file_hash}.arrow"

    r = requests.get(url)
    r.raise_for_status()

    print(
        logical_key,
        "|",
        r.status_code,
        "|",
        r.headers.get("content-type"),
        "|",
        len(r.content),
        "bytes"
    )

    path = raw_dir / f"{logical_key}.arrow"
    path.write_bytes(r.content)

    with open(path, "rb") as f:
        reader = ipc.open_stream(f)
        table = reader.read_all()

    return table.to_pandas()
```

```python
def download_arrow_from_manifest(
    manifest_obj,
    logical_key,
    output_dir
):
    file_hash = manifest_obj[logical_key]
    url = f"{arrow_base_url}{file_hash}.arrow"

    r = requests.get(url)
    r.raise_for_status()

    print(
        logical_key,
        "|",
        r.status_code,
        "|",
        r.headers.get("content-type"),
        "|",
        len(r.content),
        "bytes"
    )

    path = output_dir / f"{logical_key}.arrow"
    path.write_bytes(r.content)

    with open(path, "rb") as f:
        reader = ipc.open_stream(f)
        table = reader.read_all()

    return table.to_pandas()
```
